# for + if

### Estrutura:

Digamos que a gente esteja analisando a meta de vendas de vários funcionários de uma empresa. A meta de vendas é de 1000 reais em 1 dia.

Temos uma lista com as vendas de todos os funcionários e quero calcular qual o % de pessoas que bateram a meta.

```python
vendas = [300, 800, 400, 20, 23, 70, 90, 80, 999, 900, 880, 870, 50, 120, 300, 450, 800]
meta = 1000

total_vendas = len(vendas)
bateu_meta = []

for i in vendas:
    if i >= meta:
        bateu_meta.append(i)
        total_meta = len(bateu_meta)
        aproveitamento = total_meta / total_vendas
    else:
        print()
        

print('{} pessoas bateram a meta na empresa. Que corresponde a {} do total de funcionários.' .format(total_meta, aproveitamento))
```
