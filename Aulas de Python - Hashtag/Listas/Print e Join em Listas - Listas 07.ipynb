# Print de Listas

2 Opções:
- print "normal"
- método join -> texto.join(lista)

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
produtos = ', '.join(produtos) # 'texto'.join(lista) serve para transformar a lista em texto que será separado pelo separador que foi colocado. Ex: ','.join(lista) a lista será transformada em texto separado por vírgulas

print(produtos)
```

Lembrando do método split de strings:

lista = texto.split(separador)

```python
produtos = 'apple tv, mac, iphone x, iphone 11, IPad, apple watch, mac book, airpods'
produtos = produtos.split(',') # texto.split(',') transforma o texto em lista

print(produtos)
```
