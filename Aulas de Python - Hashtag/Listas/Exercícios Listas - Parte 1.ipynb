## 1. Faturamento do Melhor e do Pior Mês do Ano

Qual foi o valor de vendas do melhor mês do Ano?
E valor do pior mês do ano?

```python
meses = ['jan', 'fev', 'mar', 'abr', 'mai', 'jun', 'jul', 'ago', 'set', 'out', 'nov', 'dez']
vendas_1sem = [25000, 29000, 22200, 17750, 15870, 19900]
vendas_2sem = [19850, 20120, 17540, 15555, 49051, 9650]
vendas_ano = vendas_1sem + vendas_2sem # Juntei as listas de vendas dos 2 semestres em 1 de um ano

valor_maior = max(vendas_ano)
valor_menor = min(vendas_ano)

posicao_maior = vendas_ano.index(valor_maior)
mes_maior = meses[posicao_maior]
posicao_menor = vendas_ano.index(valor_menor)
mes_menor = meses[posicao_menor]

print(mes_maior, valor_maior, mes_menor, valor_menor)
```

## 2. Continuação

Agora relacione as duas listas para printar 'O melhor mês do ano foi {} com {} vendas' e o mesmo para o pior mês do ano.

Calcule também o faturamento total do Ano e quanto que o melhor mês representou do faturamento total.

Obs: Para o faturamento total, pode usar a função sum(lista) que soma todos os itens de uma lista

```python
faturamento = sum(vendas_ano)
mes_maior_representa = (valor_maior * 100) / faturamento
print('O melhor mês do ano foi {} com {} vendas.\nO pior mês do ano foi {} com {} vendas.' .format(mes_maior, valor_maior, mes_menor, valor_menor))
print('O faturamento no ano foi de R${:,.2f}. E o melhor mês do ano representou {:.2f}% do faturamento anual.' .format(faturamento, mes_maior_representa))
```

## 3. Crie uma lista com o top 3 valores de vendas do ano (sem fazer "no olho")

Dica: o método remove retira um item da lista.

```python
top3 = []

maior_valor = max(vendas_ano)
top3.append(maior_valor)
vendas_ano.remove(maior_valor)

maior_valor = max(vendas_ano)
top3.append(maior_valor)
vendas_ano.remove(maior_valor)

maior_valor = max(vendas_ano)
top3.append(maior_valor)

print(top3)
```
