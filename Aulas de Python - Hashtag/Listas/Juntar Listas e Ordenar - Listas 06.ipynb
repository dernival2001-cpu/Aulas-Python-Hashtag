# Juntar Listas, Ordenar e Cuidados Especiais

### 2 formas:

- lista1.extend(lista2)
- lista_nova = lista1 + lista2

Obs: o Método .append não junta listas, mas adiciona um valor no final da lista

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
novos_produtos = ['iphone 12', 'ioculos']

# produtos.extend(novos_produtos) e nova_lista = produtos + novos_produtos são modos de adicionar uma lista à outra
```

### Cuidado:

- [1] + [2] não é a mesma coisa que 1 + 2, então cuidado sempre com o formato dos valores na hora de fazer operações.

### Ordenar listas

lista.sort()

```python
produtos = ['apple tv', 'mac', 'iphone x', 'iphone 11', 'IPad', 'apple watch', 'mac book', 'airpods']
produtos.sort() #produtos.sort() serve para ordenar a lista em ordem alfabética e para uma lista de números, ordena-os em ordem crescente. OBS: As letras maiúsculas vêm antes da ordem alfabética
produtos.sort(reverse=True) #produtos.sort(reverse=true) serve para ordenar a lista contrário à ordem alfabética e para uma lista de números, ordena-os em ordem decrescente. OBS: As letras maiúsculas vêm antes da ordem alfabética
vendas = [1000, 1500, 15000, 20000, 270, 900, 100, 1200]
vendas.sort()
vendas.sort(reverse=True) 

print(produtos) 
print(vendas)
```
