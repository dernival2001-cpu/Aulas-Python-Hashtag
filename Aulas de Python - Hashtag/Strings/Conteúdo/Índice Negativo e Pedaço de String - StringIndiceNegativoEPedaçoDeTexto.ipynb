```python
email = 'lira@gmail.com'
nome = 'João Paulo Lira'

print(email[4:], nome[5:]) # Quando o índice é informado antes dos dois pontos e nada é informado depois: printa a partir do índice informado ate o final
print(email[:4], nome[:4]) # Quando o índice é informado depois dos dois pontos e nada é informado antes: printa a partir do índice informado, exceto ele mesmo
print(email[5:10], nome[5:11]) # Quando os dois índices são informados: printa a partir do primeiro até o índice anterior ao informado depois dos dois pontos(nesse caso o -6)
print(email[:], nome[:]) # Se os índices não forem informados, printa a Strinf inteira
```
