# Comando format

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

print('O faturamento da empresa foi de R${} e o custo foi de R${}. Portanto, o lucro foi de R${}.' .format(faturamento, custo, lucro)) # O comnado .format() transforma as variáveis fornecidas como argumentos em Strings
```
