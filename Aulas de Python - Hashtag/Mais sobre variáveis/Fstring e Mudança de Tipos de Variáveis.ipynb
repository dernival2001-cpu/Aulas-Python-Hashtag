# F-string 

```python
faturamento = 1000
custo = 500
lucro = faturamento - custo

print(f'O faturamento da empresa foi de R${faturamento} e o custo foi de R${custo}. Portanto, o lucro foi de R${lucro}.') # O f-string consiste em adicionar f antes da string, fazendo com que as variáveis dentro das chaves
```

# Mudança de Tipos de Variável

```python
faturamento = float(input('Informe o faturamento da sua empresa')) 
custo = float(input('Informe o custo da sua empresa'))
lucro = faturamento - custo

# float() serve para transformar a informação dada no input de String para float
# int() serve para transformar a informação dada no input de String para int
# str() serve para transformar a informação dada no input de qualquer tipo de variável para String

print(type(faturamento))
print(type(custo))
print(type(lucro))
```
