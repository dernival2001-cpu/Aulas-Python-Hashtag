#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
horas_trabalhadas = int(input('Quantas horas você trabalha por mês'))
ganho_por_hora = float(input('Quanto você ganha por hora trabalhada'))

# Calcule o salário bruto (horas * salario por hora)
salario_bruto = horas_trabalhadas * ganho_por_hora

# Calcule o desconto do IR (11% do salário bruto)
imposto_de_renda = salario_bruto * 0.11

# Calcule o desconto do INSS (8% do salário bruto)
inss = salario_bruto * 0.08

# Calcule o desconto do sindicato (5% do salário bruto)
sindicato = salario_bruto * 0.05

# Calcule o salário líquido (salário bruto - descontos)
descontos = imposto_de_renda + inss + sindicato
salario_liquido = salario_bruto - descontos

print(f'Seu salário bruto(sem descontos) é de R${salario_bruto}. Após serem retirados os descontos, sobra seu salário líquido, que é de R${salario_liquido}')
```

#### 14. Faça um programa que peça o tamanho de um arquivo para download (em MB) e a velocidade de um link de Internet (em Mbps), calcule e informe o tempo aproximado de download do arquivo usando este link (em minutos).

Detalhe: MB significa megabyte, Mb (com b minúsculo) significa megabit. Um megabit é 1/8 de um megabyte. 

```python
tamanho_arquivo = float(input('Digite aqui o tamanho do arquivo(MB) que deseja baixar:'))
velocidade_net = float(input('Digite aqui a velocidade(Mbps) da sua internet:'))
megabyte_p_megabit = tamanho_arquivo * 8
tempo = megabyte_p_megabit / velocidade_net
tempo_download = tempo / 60

print(f'O tempo para completar o download desse arquivo é de {tempo_download} minutos')
```
