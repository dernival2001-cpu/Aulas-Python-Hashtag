# Misturando tipos de variáveis

```python
faturamento = 1000 # A variável 'faturamento' armazena o valor inteiro 1000
custo = 500
lucro = faturamento - custo

print('O faturamento da empresa foi de R$' + str(faturamento) + '. O custo da empresa foi de R$' + str(custo) + '. Portanto, o lucro da empresa foi de R$' + str(lucro)) # O uso do str() serve para transformar o tipo da variável fornecida como argumento em String.
```
