# Tipos de variáveis em python

```python
# A variável int armazena apenas números inteiros
a_int = 1000
print('A variável a_int é do tipo' + str(type(a_int)))
```

```python
# A variável float armazena números com casas decimais
a_float = 1000.0 
print('A variável a_float é do tipo' + str(type(a_float)))
```

```python

# A variável bool ou boolean armazena valores True ou Flase
a_bool = True
print('A variável a_bool é do tipo' + str(type(a_bool)))
```

```python

# A variável string armazena textos
a_string = 'Isso é uma string'
print('A variável a_string é do tipo' + str(type(a_string)))
```
